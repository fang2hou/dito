# Polish comparison

Runs Dito's real polish prompts (mirrored from `crates/dito-core/src/prompts.rs`, general style) over dirty dictation text from `data/polish_cases.json`.

- Results are stored under `results/polish/<chip>/`; only new or stale cases run. The fingerprint includes the exact system prompt, so editing a prompt reruns the affected cases.
- Weights are downloaded with `mise run bench polish download <key>`.
- See `FINDINGS.md` for the conclusions from earlier rounds.

````python
import pandas as pd

from dito_bench.polish import benchmark
from dito_bench.polish.registry import default_polish_keys

pd.set_option("display.max_colwidth", 80)
keys = default_polish_keys()
benchmark.run(keys)
rows, warnings = benchmark.report_rows(keys)
print("\n".join(warnings))
pd.DataFrame(rows)
````

## Wordbook injection

The cases `ja-katakana-mixed` and `ja-katakana-wordbook` use the same text. The second passes the user's English terms as a wordbook, which should turn the katakana back into Latin words. That closes the loop between ASR hotwords and the polish wordbook.

````python
frame = pd.DataFrame(rows)
frame[frame["case"].str.startswith("ja-katakana")].sort_values(["case", "model"])[
    ["model", "case", "output"]
]
````
